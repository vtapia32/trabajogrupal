```python
import pandas as pd

with open('balon_de_oro.csv', 'r', encoding='utf-8', errors='ignore') as f:
    lines = f.readlines()

header_idx = 0
for i, line in enumerate(lines):
    if 'Año' in line or 'Jugador' in line:
        header_idx = i
        break

df = pd.read_csv('balon_de_oro.csv', sep=';', skiprows=header_idx)

df.columns = df.columns.str.strip()

print("¡ÉXITO! Las columnas de tu tabla ahora son:")
print(df.columns.tolist())

display(df.head())
```

```python
import altair as alt
import pandas as pd

df['Gano_Champions'] = df['Champions League'].astype(str).str.contains('Sí|Si|sí|si', case=False, na=False)

conteo = df['Gano_Champions'].value_counts().reset_index()
conteo.columns = ['Champions', 'Cantidad']
conteo['Champions'] = conteo['Champions'].replace({True: 'Ganaron Champions League', False: 'No ganaron Champions League'})

conteo['Porcentaje'] = (conteo['Cantidad'] / conteo['Cantidad'].sum()) * 100

chart = alt.Chart(conteo).mark_bar(cornerRadiusTopLeft=8, cornerRadiusTopRight=8).encode(
    x=alt.X('Champions:N', title='Condición', sort='-y'),
    y=alt.Y('Porcentaje:Q', title='Porcentaje (%) del Top 5 del Balón de Oro'),
    color=alt.Color('Champions:N', legend=None, scale=alt.Scale(range=['#2b5c8f', '#b0c4de'])),
    tooltip=['Champions', 'Cantidad', alt.Tooltip('Porcentaje:Q', format='.1f', title='Porcentaje (%)')]
).properties(
    title='Impacto de la Champions League en el Top 5 del Balón de Oro (1956-2025)',
    width=500,
    height=400
).interactive()

chart.display()
```

```python
import altair as alt
import pandas as pd

df['Posición_limpia'] = pd.to_numeric(
    df['Posición'].astype(str).str.extract(r'(\d+)')[0], errors='coerce'
)
df_ganadores = df[df['Posición_limpia'] == 1].copy()

df_ganadores['Gano_Champions'] = (
    df_ganadores['Champions League']
    .astype(str)
    .str.contains('Sí|Si|sí|si', case=False, na=False)
)

conteo = df_ganadores['Gano_Champions'].value_counts().reset_index()
conteo.columns = ['Champions', 'Cantidad']
conteo['Champions'] = conteo['Champions'].replace(
    {True: 'Ganaron Champions League', False: 'No ganaron Champions League'}
)

total_ganadores = conteo['Cantidad'].sum()
conteo['Porcentaje'] = (conteo['Cantidad'] / total_ganadores) * 100

print(f'Total de ganadores analizados: {total_ganadores}')
display(conteo)

chart = (
    alt.Chart(conteo)
    .mark_bar(size=60)
    .encode(
        x=alt.X(
            'Champions:N',
            title='Condición del Ganador del Balón de Oro',
            sort='-y',
        ),
        y=alt.Y(
            'Porcentaje:Q', title='Porcentaje (%) de los Ganadores del Balón de Oro'
        ),
        color=alt.Color('Champions:N', legend=None),
        tooltip=['Champions', 'Cantidad', alt.Tooltip('Porcentaje:Q', format='.1f', title='Porcentaje (%)')]
    )
    .properties(
        title=(
            'Porcentaje de Ganadores del Balón de Oro que conquistaron la'
            ' Champions League'
        ),
        width=450,
        height=350,
    )
    .interactive()
)

chart.display()
```

```python
import altair as alt
import pandas as pd

df['Año_num'] = pd.to_numeric(df['Año'].astype(str).str.extract(r'(\d+)')[0], errors='coerce')
df['Posición_limpia'] = pd.to_numeric(df['Posición'].astype(str).str.extract(r'(\d+)')[0], errors='coerce')

df_filtrado = df[(df['Posición_limpia'] == 1) & (df['Año_num'] >= 1956) & (df['Año_num'] <= 1999)].copy()

df_filtrado['Gano_Champions'] = df_filtrado['Champions League'].astype(str).str.contains('Sí|Si|sí|si', case=False, na=False)

conteo = df_filtrado['Gano_Champions'].value_counts().reset_index()
conteo.columns = ['Champions', 'Cantidad']
conteo['Champions'] = conteo['Champions'].replace({True: 'Ganaron Champions League', False: 'No ganaron Champions League'})

total_ganadores = conteo['Cantidad'].sum()
conteo['Porcentaje'] = (conteo['Cantidad'] / total_ganadores) * 100

print(f"Total de ganadores analizados entre 1956 y 1999: {total_ganadores}")
display(conteo)

base = alt.Chart(conteo).encode(
    theta=alt.Theta(field="Cantidad", type="quantitative"),
    color=alt.Color(field="Champions", type="nominal", title="Condición del Ganador")
)

pie = base.mark_arc(outerRadius=130, innerRadius=65).encode(
    tooltip=['Champions', 'Cantidad', alt.Tooltip('Porcentaje:Q', format='.1f', title='Porcentaje (%)')]
)

chart_pie = pie.properties(
    title='Ganadores del Balón de Oro (1956-1999): Relación con la Champions League',
    width=500,
    height=400
).interactive()

chart_pie.display()
```
